```python
data = {
    "customer_id": 101,
    "customer_name": "Anurag",
    "orders": [
        {"order_id": 1, "amount": 500},
        {"order_id": 2, "amount": 800}
    ]
}
```

```python
records=[]
for orders in data['orders']:
    print(orders)
    records.append({
        "customer_id":data["customer_id"],
        "customer_name":data["customer_name"],
        "order_id":orders["order_id"],
        "amount":orders["amount"]
    })
print(records)
display(records)
```
